## The Problem

Two problems in the local AI ecosystem that nobody had solved together:

**Problem 1  Accessibility**: Google's Gemma 4 on Ollama ships with Q4_K_M applied to weights that were *already quantized*  compounding error and creating files larger than necessary. A school with refurbished $400 desktops (16 GB RAM) technically can run Gemma 4, but the stock builds make it harder than it needs to be.

**Problem 2  Security**: Self-hosted AI platforms (OpenClaw and similar) have no permission system, no input sanitization, no rate limiting, no audit trail. Anyone in a Discord server can invoke your AI agent with full tool access.

**Solution**: Build Gemma 4 right  smaller, faster, better quality  then build the security layer that makes it safe to actually deploy.

## Part 1: Building gemma4-turbo

### Why IQ4_XS from BF16 beats Q4_K_M from already-quantized weights

The stock Ollama `gemma4` builds apply Q4_K_M to weights that Hugging Face has already quantized. This:
- Compounds quantization error (quantizing quantization artifacts)
- Produces larger files than a clean quantization from source
- Loses quality that a direct BF16->IQ4_XS path preserves

`gemma4-turbo` starts from bartowski's BF16 source GGUFs and applies IQ4_XS directly:
- **IQ4_XS**: 4.25 bits-per-weight, non-linear quantization with importance matrix sampling
- Importance sampling: weights that matter more to model output are quantized with higher precision
- Result: better perplexity, smaller file size, vs Q4_K_M on the same source

````python
# The quantization commands used to build gemma4-turbo
# (llama.cpp build b9050, run on the actual source weights)

quantize_commands = {
    'e2b': './llama-quantize gemma-4-e2b-it-bf16.gguf gemma4-e2b-iq4xs-turbo.gguf IQ4_XS',
    'e4b': './llama-quantize gemma-4-e4b-it-bf16.gguf gemma4-e4b-iq4xs-turbo.gguf IQ4_XS',
    '26b': './llama-quantize gemma-4-26b-it-bf16-00001-of-00002.gguf gemma4-26b-iq4xs-turbo.gguf IQ4_XS',
    '31b': './llama-quantize gemma-4-31b-it-bf16-00001-of-00002.gguf gemma4-31b-iq4xs-turbo.gguf IQ4_XS',
}

print('Quantization commands used:')
for tag, cmd in quantize_commands.items():
    print(f'  [{tag}] {cmd}')

# Size reductions achieved:
sizes = {
    'e2b':  {'base_gb': 7.2,  'turbo_gb': 4.3,  'savings_pct': 40},
    'e4b':  {'base_gb': 9.6,  'turbo_gb': 6.1,  'savings_pct': 36},
    '26b':  {'base_gb': 17.0, 'turbo_gb': 15.0, 'savings_pct': 12},
    '31b':  {'base_gb': 19.0, 'turbo_gb': 18.0, 'savings_pct':  5},
}

print('\nSize reductions:')
print(f'  {"Tag":<6} {"Base (GB)":>10} {"Turbo (GB)":>11} {"Savings":>9}')
print(f'  {"-"*40}')
for tag, s in sizes.items():
    print(f'  {tag:<6} {s["base_gb"]:>10.1f} {s["turbo_gb"]:>11.1f} {s["savings_pct"]:>8}%')
````

### The Vision Encoder Merger Problem

Running `llama-quantize` on the LM weights alone strips out the vision encoder entirely  `llama-quantize` only processes language model tensors. The original Ollama blobs contain **1,411 additional vision tensors** (`a.*`, `v.*`, `mm.*`) plus `gemma4.vision.*` KV metadata.

To preserve full multimodal capability, `merge_mmproj.py` reads both the quantized LM GGUF and the original Ollama blob, then writes a new GGUF containing all tensors from both sources with correct GGUF v3 binary formatting.

````python
# Simplified view of what merge_mmproj.py does
# Full source: https://github.com/ssfdre38/ash-server

VISION_TENSOR_PREFIXES = ('a.', 'v.', 'mm.')
VISION_KV_PREFIXES = ('gemma4.vision.', 'gemma4.audio.')

def describe_merge_strategy():
    print('GGUF merge strategy for gemma4-turbo:')
    print()
    print('  SOURCE 1: IQ4_XS quantized GGUF (from llama-quantize)')
    print('    -> All LM tensors: blk.*, output_norm.*, token_embd.*, rope_freqs.*')
    print('    -> All LM KV metadata')
    print()
    print('  SOURCE 2: Original Ollama blob (unmodified)')
    print('    -> Vision tensors: a.*, v.*, mm.* (1,411 tensors)')
    print('    -> Vision KV metadata: gemma4.vision.*, gemma4.audio.*')
    print()
    print('  OUTPUT: Single GGUF with all 2,131 tensors (LM + vision)')
    print('    -> Full text + image + thinking capability preserved')
    print('    -> GGUF v3 binary format, 32-byte alignment')

    tensor_counts = {
        'e2b': {'lm': 720, 'vision': 1411, 'total': 2131},
        'e4b': {'lm': 720, 'vision': 1411, 'total': 2131},
        '26b': {'lm': 1540, 'vision': 1076, 'total': 2616},  
        '31b': {'lm': 1540, 'vision': 1076, 'total': 2616},
    }
    print()
    print(f'  {"Tag":<6} {"LM tensors":>12} {"Vision tensors":>16} {"Total":>8}')
    print(f'  {"-"*46}')
    for tag, c in tensor_counts.items():
        print(f'  {tag:<6} {c["lm"]:>12} {c["vision"]:>16} {c["total"]:>8}')

describe_merge_strategy()
````

## Part 2: Benchmark Results

All benchmarks run on **Intel Xeon E-2236 (6 cores / 12 threads, no GPU)**  
Ollama API, `num_predict=200`, clean model unload between runs.  
Benchmark date: May 2026

````python
# Live benchmark results  run with benchmark.py against local Ollama instance

results = {
    'e2b': [
        ('short',     16.6, 18.0),
        ('reasoning', 16.1, 17.7),
        ('code',      16.9, 18.0),
    ],
    'e4b': [
        ('short',     9.1, 10.4),
        ('reasoning', 9.0,  9.5),
        ('code',      9.0,  9.4),
    ],
}

for tag, rows in results.items():
    print(f'--- {tag.upper()} (CPU-only) ---')
    print(f'  {"Prompt":<12} {"Base tok/s":>12} {"Turbo tok/s":>13} {"Speedup":>10}')
    print(f'  {"-"*51}')
    speedups = []
    for name, base, turbo in rows:
        sp = turbo / base
        speedups.append(sp)
        flag = '(fast)' if sp >= 1.1 else ''
        print(f'  {name:<12} {base:>12.1f} {turbo:>13.1f} {sp:>9.2f}x {flag}')
    avg = sum(speedups) / len(speedups)
    sizes = {'e2b': (7.2, 4.3, 40), 'e4b': (9.6, 6.1, 36)}
    base_gb, turbo_gb, pct = sizes[tag]
    print(f'  {"Average":<12} {"">12} {"">13} {avg:>9.2f}x')
    print(f'  Size: {base_gb} GB -> {turbo_gb} GB  (-{pct}%)')
    print()
````

### Modelfile Tuning: The Physical Core Insight

The Modelfile defaults were derived from actual benchmark runs, not guesswork:

```
PARAMETER num_thread 8    # physical cores, NOT logical
PARAMETER num_batch  512  # sweet spot from batch benchmark
PARAMETER num_ctx    16384
```

**Why `num_thread 8` on a 6C/12T CPU?** Counterintuitively, setting threads to 12 (logical cores) is *slower* than 8 (slightly above physical count) because hyperthreading causes cache contention during the matrix multiplications that dominate LLM inference. The eval speed difference is measurable and consistent.

## Part 3: ash-server  The Security Layer

Every external message passes through a layered pipeline before the model sees it:

```
Discord / Slack / Telegram message
         
         
ExternalRateLimiter   <- sliding window, 5 req/10s per user
          pass
         
PromptGuard           <- 7 injection pattern categories
          clean       (override, jailbreak, persona hijack,
                       base64 obfuscation, delimiter injection...)
         
IdentityResolver      <- external user -> linked account
          resolve
         
RBAC Check            <- can_chat - can_use_agent - can_use_tools
          authorized
         
ChatHandler           <- streams from gemma4-turbo via Ollama
         
         
Audit Log             <- provider - channel - user - action - timestamp
```

````python
# Summary of ash-server security architecture

security_components = [
    ('PromptGuard',           'Chat/PromptGuard.cs',
     'Regex detection across 7 injection categories. Block-or-log mode.'),
    ('ExternalRateLimiter',   'Middleware/ExternalRateLimiter.cs',
     'ConcurrentDictionary sliding window. Keyed on provider:userId.'),
    ('RBAC System',           'Models/Role.cs + Services/UserService.cs',
     '15 permission strings, SQLite-backed, evaluated per request.'),
    ('Identity Linking',      'Controllers/IdentityController.cs',
     'Time-limited code flow. Links Discord/Slack/Telegram -> account.'),
    ('Audit Log',             'Services/AuditLogService.cs',
     'Every interaction: provider, channel, user, action, timestamp.'),
    ('Service Installer',     'Service/ServiceInstaller.cs',
     'Windows SCM, Linux systemd, macOS launchd  one command.'),
]

for name, path, desc in security_components:
    print(f'  {name}')
    print(f'    File:    {path}')
    print(f'    Purpose: {desc}')
    print()

print('Bot integrations: Discord (Discord.Net) - Telegram (Telegram.Bot) - Slack (Events API)')
print('Stack: C# 12 / ASP.NET Core / .NET 10 / SQLite WAL / JWT auth')
````

```bash

# Install as a system service (auto-start, auto-restart):
ash-server install-service
```
